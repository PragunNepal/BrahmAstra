```python
# 1. SETUP & CLEANUP
# ==========================================
%load_ext autoreload
%autoreload 2

import os
import brahmastra as bm

print("🧹 CLEANUP: Wiping old data...")
os.system("rm -rf output.nbody_* halo_* ionz_out/ bispec_out/ simulation.log *.mp4 *.png input.bispec dvisukta_param.txt c_data*")

# 2. THE PIPELINE COMPUTE PHASE
# ==========================================
print("\n🚀 LAUNCHING THE BRAHMASTRA PIPELINE...")

# Using a tiny 16^3 grid and spanning z=100 down to z=1
test_setup = {
    "grid_size": 16,
    "z_start": 100.0,
    "z_end": 1.0,
    "num_redshifts": 50 
}

sim = bm.BrahmAstraPipeline(params=test_setup)
sim.run_full_simulation()


# 3. THE VISUALIZATION PHASE
# ==========================================
print("\n🎨 INITIATING COSMO-VIS RENDERERS...")
vis = bm.CosmoVis()

# --- Test A: DviSukta Bispectrum ---
print("\n[Test A] DviSukta Plots")
# 1. Default (Should grab z=1.0 automatically)
vis.plot_bispectrum(output_filename="Test_Bispec_Default_z1.png")
# 2. Custom Redshift (Will find the closest match to z=45.0)
vis.plot_bispectrum(redshift=45.0, output_filename="Test_Bispec_Custom_z45.png")


# --- Test B: 3D N-body (Dark Matter) ---
print("\n[Test B] 3D N-body Animation")
vis.animate_nbody(
    output_filename="Test_Nbody_3D.mp4",
    show_grid=True, grid_color='green',  # Testing the custom grid
    rotate=True, rotation_speed=0.5,     # Fast rotation
    pause_per_redshift_sec=1.0,          # 1 second freeze per redshift
    freeze_end_sec=2.0,                  
    zoom_in=True, zoom_factor=1.5
)


# --- Test C: 3D Halos (FoF) ---
print("\n[Test C] 3D Halo Animation")
vis.animate_halos(
    output_filename="Test_Halos_3D.mp4",
    show_grid=False,                     # No grid for this one
    rotate=False,                        # Testing static camera
    pause_per_redshift_sec=0.5,
    freeze_end_sec=1.0
)


# --- Test D: 3D ReionYuga (HI Map) ---
print("\n[Test D] 3D PyVista HI Map")
vis.animate_hi_map(
    output_filename="Test_HImap_3D.mp4",
    rotate=True, rotation_speed=0.4,
    pause_per_redshift_sec=1.0,
    freeze_end_sec=2.0
)


# --- Test E: 2D ReionYuga Fallback ---
print("\n[Test E] 2D Matplotlib HI Map (Fallback)")
vis.animate_hi_map_2d(
    output_filename="Test_HImap_2D_Fallback.mp4",
    pause_per_redshift_sec=1.0,
    freeze_end_sec=2.0
)

print("\n🎉 ALL TESTS COMPLETE! Check your directory for the MP4s and PNGs!")
```

```python
import brahmastra as bm
vis = bm.CosmoVis()

# Let's pick 3 specific redshifts from your run:
# (Early universe, middle universe, late universe)
target_redshifts = [20.0, 10.4, 3.0]

# Let's pick 3 different k_ratio (k2/k1) values to look at different triangle configurations:
target_k_ratios = [0.5, 1.0, 1.5]

print("🎨 Generating Bispectrum Gallery...")

for z in target_redshifts:
    for k in target_k_ratios:
        # Dynamically name the output files so they don't overwrite each other
        custom_name = f"Bispec_z{z}_k{k}.png"
        
        # The visualizer will automatically find the closest matching data!
        vis.plot_bispectrum(
            redshift=z, 
            k_ratio=k, 
            output_filename=custom_name
        )
        
print("\n🎉 Done! Check your folder for the new gallery of PNG files.")
```
